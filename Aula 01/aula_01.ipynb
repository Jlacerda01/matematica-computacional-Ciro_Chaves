Cenário de aplicação: Consideremos um Sistema de Controle de Acesso baseado em cargos, ou um *RBAC - Role-Based Acess Control -* como sendo a nossa primeira aplicação prática. Consideremos o uso de *sets*, que diferentemente das listas em Python são estruturas desordenadas de itens únicos, ou seja, sem duplicatas. 

Consideremos, primeiro, o uso da estrutura *set* para definir os conjuntos de usuários com especificações bem definidas em nosso exemplo:

```python
desenvolvedores = {'Ana', 'Bruno', 'Carlos', 'Diana'}
analistas_banco = {'Carlos', 'Diana', 'Eduardo'}
administradores = {'Ana', 'Eduardo', 'Fernanda'}
```

Consideremos o seguinte conjunto de regras e especificações que precisamos aplicar:
* Preciso definir quem possui acesso aos servidores de homologação. Os servidores de homologação podem ser acessados por desenvolvedores ou analistas de banco
* Preciso "separar" aqueles colaboradores que são devs e também administradores, para participar de um projeto
* Preciso determinar aqueles que são desenvolvedores, mas não possuem acesso ao banco.

```python

acesso_homologacao = desenvolvedores | analistas_banco
print(f'Acesso à Homologação: {acesso_homologacao}')

dev_admins = desenvolvedores & administradores
print(f'Devs com privilégios de Admin: {dev_admins}')

apenas_devs = desenvolvedores - analistas_banco
print(f'Devs que não possuem acesso ao banco: {apenas_devs}')
```

Neste outro exemplo, vamos utilizar as *bitmasks* para realizar o mapeamento de permissões em um sistema computacional. Consideremos o seguinte mapeamento de permissões:

* 0b001 (1) -> Ler
* 0b010 (2) -> Escrever
* 0b100 (4) -> Executar

Considere o perfil do usuário Carlos, que deve ser de permissçao para ler e executar neste sistema: Como eu posso configurar o sistema para que ele possua essas atribuições?

```python
PERMISSAO_LER = 0b001
PERMISSAO_ESCREVER = 0b010
PERMISSAO_EXECUTAR = 0b100

perfil_carlos = PERMISSAO_LER | PERMISSAO_EXECUTAR # Representação: 101
```

Um exemplo de verificação de permissão específica:

```python
if perfil_carlos & PERMISSAO_LER:
    print('Permissão para leitura.')
else:
    print('Acesso negado para leitura.')

if perfil_carlos & PERMISSAO_ESCREVER:
    print('Permissão para escrita.')
else:
    print('Acesso negado para escrita.')

if perfil_carlos & PERMISSAO_EXECUTAR:
    print('Permissão para executar.')
else:
    print('Permissão para execução.')
```

## Exercício proposto 2

Considere a flag de verificação de status de três servidores operando como se mostra, indicando que os servidores 1 e 3 estão em operação, e o servidor 2 está offline:

```python
status_servidores = 0b101
```

Faça o que se pede:
1) Crie uma máscara e aplique uma operação de bitmask para modificar o status dos servidores para manter os três ativos. Aplique a máscara e mostre na tela o resultado antes e depois, verificando que o flag correspondente foi *setado* por meio de uma condicional *if*.

```python
# Dados de log brutos
log_bruto_frontend = [101, 102, 103, 104, 105, 101, 102, 103]
log_bruto_backend = [103, 104, 106, 107, 108, 103, 106]
log_bruto_devops = [103, 105, 108, 109, 110, 109, 103]

# 1. Transformar listas em conjuntos para eliminar duplicatas
frontend = set(log_bruto_frontend)
backend = set(log_bruto_backend)
devops = set(log_bruto_devops)

print("Conjunto Frontend:", frontend)
print("Conjunto Backend:", backend)
print("Conjunto DevOps:", devops)

# 2. Usuários Fullstack (intersecção entre Frontend e Backend)
users_fullstack = frontend & backend
print("\nUsers Fullstack (Front & Back):", users_fullstack)

# 3. Usuários que NÃO estão no Frontend (União de Back e DevOps minus Frontend)
users_no_front = (backend | devops) - frontend
print("Users sem Frontend:", users_no_front)

# 4. Usuários EXCLUSIVAMENTE Frontend (Diferença de conjuntos)
users_xor_front = frontend - (backend | devops)
print("Users exclusivamente Frontend:", users_xor_front) 

#+------++-+-+-+--------+-++-+-----------+-+-+-+-+------------------+-+-+-+-+-+-+

# Status inicial: Servidor 3 e Servidor 1 ativos (101 em binario = 5 em decimal)
status_servidores = 0b101

# 1. Máscara de bits para ativar o Servidor 2 (posição do segundo bit = 0b010 ou 1 << 1)
mascara = 0b010

# Exibição antes da modificação
print(f"Status Antes:  {bin(status_servidores)} (Decimal: {status_servidores})")

# Aplicação do Bitmask (Operador OR '|' para SETAR o bit do servidor 2)
status_servidores_atualizado = status_servidores | mascara

# Exibição depois da modificação
print(f"Status Depois: {bin(status_servidores_atualizado)} (Decimal: {status_servidores_atualizado})")

# Verificação com a condicional if (Operador AND '&' para checar a flag)
if (status_servidores_atualizado & mascara) == mascara:
    print("\n[VERIFICAÇÃO]: O Servidor 2 foi ativado com sucesso! Todos os 3 servidores estão operacionais.")
else:
    print("\n[VERIFICAÇÃO]: O Servidor 2 permanece offline.")
```
